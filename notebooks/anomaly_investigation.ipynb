```python
import pandas as pd

cleaned_df = pd.read_csv(
    "/Users/rajpoot/Desktop/customer-intelligence/data/processed/cleaned_transactions.csv"
)

cleaned_df["InvoiceDate"] = pd.to_datetime(
    cleaned_df["InvoiceDate"]
)

cleaned_df["CustomerID"] = cleaned_df["CustomerID"].astype(int)

print("Dataset shape:", cleaned_df.shape)
```

```python
customers_to_investigate = [16446, 12346, 14646]

investigation_data = cleaned_df[
    cleaned_df["CustomerID"].isin(customers_to_investigate)
].copy()

print("Investigation rows:", len(investigation_data))

display(investigation_data.head())
```

```python
customers_to_investigate = [16446, 12346, 14646]

investigation_data = cleaned_df[
    cleaned_df["CustomerID"].isin(customers_to_investigate)
].copy()

print("Investigation rows:", len(investigation_data))

display(investigation_data.head())
```

```python
invoice_report = (
    investigation_data.groupby(["CustomerID", "InvoiceNo"])
    .agg(
        invoice_date=("InvoiceDate", "min"),
        quantity=("Quantity", "sum"),
        revenue=("Revenue", "sum"),
        unique_products=("StockCode", "nunique"),
    )
    .reset_index()
)

display(
    invoice_report.sort_values(
        ["CustomerID", "revenue"],
        ascending=[True, False]
    )
)
```

```python
customer_16446 = investigation_data[
    investigation_data["CustomerID"] == 16446
].copy()

print("Transaction rows:", len(customer_16446))
print("Unique invoices:", customer_16446["InvoiceNo"].nunique())
print("Unique products:", customer_16446["StockCode"].nunique())

display(
    customer_16446[
        [
            "InvoiceNo",
            "InvoiceDate",
            "StockCode",
            "Description",
            "Quantity",
            "UnitPrice",
            "Revenue",
        ]
    ].sort_values("InvoiceDate")
)
```

```python
customer_12346 = investigation_data[
    investigation_data["CustomerID"] == 12346
]

display(customer_12346)
```

```python
customer_report = (
    investigation_data.groupby("CustomerID")
    .agg(
        invoice_count=("InvoiceNo", "nunique"),
        transaction_lines=("InvoiceNo", "count"),
        total_quantity=("Quantity", "sum"),
        total_revenue=("Revenue", "sum"),
        unique_products=("StockCode", "nunique"),
        first_purchase=("InvoiceDate", "min"),
        last_purchase=("InvoiceDate", "max"),
        country=("Country", "first"),
    )
)

customer_report.index = customer_report.index.astype(int)

print(customer_report)
```
